## Funktionen
Eine Funktion in Python kann als Unterprogramm aufgefasst werden, das ggf. einen Rückgabewert liefert. ```sin()``` und ```exp()```sind zum Beispiel Funktionen. 

In Python können mit dem Schlüsselwort ```def``` auch eigene Funktionen erstellt werden. Der folgende Code-Block liefert ein Beispiel.

````python
# Funktion definieren
def myMathFun(a,b):
    import math as m
    y = 2*m.cos(a*m.pi+b)
    return y

# Funktion verwenden
print(myMathFun(2,3))
````

## Klassen
Python unterstützt *objektorientierte* Programmierung. Im Folgenden soll kurz eine Klasse definiert und verwendet werden. Wichtig ist aber nur, dass Sie mit diesem Konzept vertraut sind, nicht, dass Sie selbst Klassen implementieren können.

````python
# Klasse definieren
class id_gas:
    """
    Klasse für ein ideales Gas
    """
    R    = 8.31           # Gaskonstante in J/(K*mol)
    K_B  = 1.38e-23       # Boltzmannkonstante in J/K

    # Konstruktor
    def __init__(self,n,T,V):
        """
        Initialisiert ein ideales Gas unter Angabe von
        n Stoffmenge
        T Temperatur
        V Volumen
        """
        self.__stoffm = n  # Stoffmenge in mol
        self.__temp   = T  # Temperatur in K
        self.__Vol    = V  # Volumen in m^3
        self.__p      = n*self.R*T/V    # Druck in Pa

    # Zustandsabfrage
    def getState(self):
        state = [self.__stoffm, self.__temp,self.__Vol,self.__p]
        return state

    # Isotherme Expansion
    def isoExp(self,V_n):
        self.__Vol = V_n
        self.__p   = self.__stoffm*self.R*self.__temp/self.__Vol
        print("Druck: %8.2f Pa, Temperatur: %8.2f K, Volumen: %8.4f m^3."%(self.__p,self.__temp,self.__Vol))  
        state = self.getState()
        return state

# Klassenvariablen
print("Die Gaskonstante ist %f J/(K*mol)"%(id_gas.R))

# Klasse verwenden: Instanz erzeugen
myG = id_gas(2,300,.002)


# Zustand abfragen
zustand = myG.getState()
print(zustand)


# Isotherme Expansion
zustand = myG.isoExp(10)
print(zustand)
````

# NumPy und Matplotlib
Das Modul NumPy stellt etliche Methoden für umfangreichere Berechnungen zur Verfügung. Das folgende Beispiel berechnet die Funktion
$$
    f(x) = \frac{\sin(\pi x)}{\pi x}
$$
für $0,01<x<10$.

````python
import numpy as np

x = np.linspace(.1,10,1000)
f = np.sin(np.pi*x)/(np.pi*x)
````

## Einfache Diagramme
Zur Auftragung der Daten in einem Diagramm kann das Modul Matplotlib verwendet werden. Hiervon wird das Untermodul pyplot benötigt.

````python
import matplotlib.pyplot as plt


plt.plot(x,f)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.show()
````

## Kurven im Raum
Nachfolgend wird eine Schraubenlinie im Raum gezeichnet. Für diese gilt:

$$\begin{aligned}
    x   &= R\cos t\\
    y   &= R\sin t\\
    z   &= vt.
\end{aligned}$$

````python
import numpy as np
import matplotlib.pyplot as plt

R   = 6
v_0 = 5
w   = 3

fig = plt.figure()
ax  = fig.add_subplot(projection='3d')

t   = np.linspace(0,2*np.pi,500)

x   = R*np.cos(w*t)
y   = R*np.sin(w*t)
z   = v_0*t

ax.plot(x,y,z)

plt.show()
````

## Oberflächen
Nachfolgend ist die Funktion 
$$
    z=f(x,y)=5-x^2-y^2

````python
import numpy as np
import matplotlib.pyplot as plt

ax = plt.figure().add_subplot(projection='3d')

x = np.linspace(-5,5,200)
y = x
x,y = np.meshgrid(x,y)
z = 5-x**2-y**2

ax.plot_surface(x,y,z)

plt.show()
````

## Kugeloberfläche
Als Beispiel für parametrische Oberflächen. $r$ wird konstant gehalten, $\vartheta$ und $\varphi$ werden variiert. Dabei entstehen Auschschnitte aus einer Kugeloberfläche.

Zur Erinnerung:
$$
    x = r\sin\vartheta\cos\varphi\\
    y = r\sin\vartheta\sin\varphi\\
    z = r\cos\vartheta
$$

Der nachfolgende Code zeichnet den Ausschnitt aus einer Kugeloberfläche für 
$$\begin{aligned}
    0&\leq \varphi\leq \frac{\pi}{2}, &0&\leq \vartheta\leq \frac{\pi}{2}. 
\end{aligned}$$

````python
ax = plt.figure().add_subplot(projection='3d')

r   = 5
phi = np.linspace(0,.5*np.pi,200)
th  = np.linspace(0,.5*np.pi,200)

phi, th = np.meshgrid(phi,th)

x = r*np.sin(th)*np.cos(phi) 
y = r*np.sin(th)*np.sin(phi)
z = r*np.cos(th)

ax.plot_surface(x,y,z)
ax.set_xlim(-r,r)
ax.set_ylim(-r,r)
ax.set_zlim(-r,r)
ax.axis.__eq__(True)

plt.show()
````

$$\begin{aligned}
    2 &\leq r \leq 5\\
    0 &\leq\varphi \leq 2\pi \\
    \vartheta &= \frac{\pi}{6} =\text{konst.} 
\end{aligned}$$
